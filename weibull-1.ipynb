```python
# import libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# weibull
def weibull_cdf(t, t0, T, b):
    """
    Weibull cumulative distribution function:
    F(t) = 1 - exp(-((t - t0)/(T-t0)^b)

    F(t) = 0 for t < t0
    """
    F = np.zeros_like(t, dtype=float)

    valid = t >= t0
    F[valid] = 1 - np.exp(-((t[valid] - t0) / (T-t0)) ** b)

    return F


# Parameters
t0 = 0      # location / threshold
T_1 = 20000    # characteristic life v1
T_2 = 39000    # characteristic life v2

# Time axis
t_1 = np.linspace(0, 30000, 301)
t_2 = np.linspace(0, 60000, 601)

# Shape parameters to compare
b_values = [0.5, 1.0, 2.0, 5.0, 10.0]

plt.figure(figsize=(8, 5))

for b in b_values:
    F_1 = weibull_cdf(t_1, t0, T_1, b)
    plt.plot(t_1, F_1, label=f"b = {b}")

plt.title("Weibull Distribution")
plt.xlabel("Flight Hours")
plt.ylabel("failure probability")
plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()

# different characteristic life
plt.figure(figsize=(8, 5))

for b in b_values:
    F_2 = weibull_cdf(t_2, t0, T_2, b)
    plt.plot(t_2, F_2, label=f"b = {b}")

plt.title("Weibull Distribution")
plt.xlabel("Flight Hours")
plt.ylabel("failure probability")
plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()
```

```python
# visualize Weibull functions for Appendices G, H, I, J

# Time axis
t_3 = np.linspace(0, 35000, 351)


# parameters of different appendices G, H, I, J
appendices = ["G", "H", "I", "J"]
t0 = [19858, 19857, 19307, 19772]
b = [21.474, 0.4, 11.318, 6.214]
T_G = [20003, 34286, 20144, 20069]


plt.figure(figsize=(8, 5))
for i in range(4):
    F_i = weibull_cdf(t_3, t0[i], T_G[i], b[i])
    plt.plot(t_3, F_i, label=f"Appendix {appendices[i]}")

plt.title("Weibull Distribution")
plt.xlabel("Flight Hours")
plt.ylabel("failure probability")
plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()
```
